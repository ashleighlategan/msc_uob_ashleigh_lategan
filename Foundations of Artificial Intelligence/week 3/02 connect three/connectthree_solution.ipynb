### Objective
You should write or adapt the minimax algorithm from the noughts and crosses example to work for connect three. The state space is much larger due to the size of the board, so the basic brute force version of the algorithm is unlikely to be computationally efficient enough to actually play against. Hence you will be required to implement some optimisations, for example:
* Alpha-beta pruning – see the noughts and crosses example.
* Cache results for board states – multiple paths can lead to the same state so you shouldn't have to solve these twice. You can use a dictionary to cache results efficiently.
* Optimise for symmetry – the board is symmetric around the centre column, so you can cut down the number of necessary states by assuming that optimal play will look the same but adjusted for reflection when you are creating the cache.
* Table lookups – like caching results but saving permanently, you can precompute a certain number of end-game board positions and store the results (e.g. calculate in one cell and then use in another).
* Depth limit – a hard cap on the recursion limit will hugely improve performance at the cost of optimality. If the recursion limit is reached you can simply return a default value like 0, or...
* Evaluation functions – rather than returning a default value when you hit the recursion limit, you can use a fast heuristic function to evaluate the strength of a board position for the given player. What you choose is up to you! Be creative!

### Solution overview

The plan, building on the noughts-and-crosses minimax example:

1. **A lightweight board** (`Board`) that the search can modify and undo cheaply. Copying the `Connect` object at every node would be very slow, so the search works on plain Python lists and only uses `env` to actually play moves.
2. **Minimax with alpha-beta pruning.** `'o'` is the maximiser (because `env.act()` returns rewards for `'o'`) and `'x'` is the minimiser.
3. **A cache (transposition table)** keyed on the board, so positions reached by different move orders are only solved once.
4. **Symmetry**: a board and its mirror image have the same value, so both share one cache entry.
5. **Move ordering**: try the centre columns first (and the best move remembered in the cache). Good moves first = more alpha-beta cut-offs.
6. **Depth limit + evaluation function**: not needed to solve 3×5 Connect Three, but essential for full Connect Four.
7. **Table lookup**: solve the whole game once, save the cache to disk, and load it next time, so the AI then moves instantly.

**Scoring.** A win for `'o'` scores `+(1 + empty squares left)`, a win for `'x'` scores the negative of that, and a draw scores 0. Adding the number of empty squares means the AI prefers to win *quickly* and to lose *slowly*. It also means the score depends only on the board, not on how deep we are in the search, so it is safe to cache.

```python
import time
import pickle
import random
import numpy as np
import connect


class Board:
    """Fast, mutable board used inside the search (row 0 = bottom, as in Connect)."""

    def __init__(self, num_rows=3, num_cols=5, num_connect=3):
        self.rows, self.cols, self.k = num_rows, num_cols, num_connect
        self.cells = [[" "] * num_cols for _ in range(num_rows)]
        self.heights = [0] * num_cols          # number of disks in each column
        self.empty = num_rows * num_cols       # empty squares left
        self.lines = self._all_lines()         # every possible line of k squares
        # For each square, the lines passing through it (so we only check those after a move)
        self.lines_through = {(r, c): [] for r in range(num_rows) for c in range(num_cols)}
        for line in self.lines:
            for sq in line:
                self.lines_through[sq].append(line)

    @classmethod
    def from_env(cls, env):
        b = cls(env._num_rows, env._num_cols, env._num_connect)
        for r in range(b.rows):
            for c in range(b.cols):
                if env.grid[r, c] != " ":
                    b.cells[r][c] = str(env.grid[r, c])
                    b.heights[c] += 1
                    b.empty -= 1
        return b

    def _all_lines(self):
        lines = []
        for r in range(self.rows):
            for c in range(self.cols):
                for dr, dc in [(0, 1), (1, 0), (1, 1), (1, -1)]:  # right, up, up-right, up-left
                    line = [(r + i * dr, c + i * dc) for i in range(self.k)]
                    if all(0 <= rr < self.rows and 0 <= cc < self.cols for rr, cc in line):
                        lines.append(line)
        return lines

    def legal_moves(self):
        return [c for c in range(self.cols) if self.heights[c] < self.rows]

    def play(self, col, player):
        r = self.heights[col]
        self.cells[r][col] = player
        self.heights[col] += 1
        self.empty -= 1
        return r

    def undo(self, col):
        self.heights[col] -= 1
        self.cells[self.heights[col]][col] = " "
        self.empty += 1

    def is_win_at(self, r, c, player):
        """Did the disk just placed at (r, c) complete a line for `player`?"""
        cells = self.cells
        return any(all(cells[rr][cc] == player for rr, cc in line)
                   for line in self.lines_through[(r, c)])

    def key(self):
        """Canonical cache key: the board or its mirror image, whichever sorts first."""
        k = "|".join("".join(row) for row in self.cells)
        k_mirror = "|".join("".join(reversed(row)) for row in self.cells)
        return (k, False) if k <= k_mirror else (k_mirror, True)

    def __str__(self):
        return str(np.array(self.cells)[::-1])
```

#### The evaluation function

When the depth limit is reached we can't see the end of the game, so we estimate how good the position is for `'o'`. For every possible line of `k` squares:

* if it contains **only `'o'` disks** (and blanks), it is still "alive" for `'o'`, so add `10^(n-1)` where `n` is the number of `'o'` disks in it;
* if it contains **only `'x'` disks**, subtract the same amount;
* if it contains both colours it is dead and counts for nothing.

So two-in-a-line with the third square free is worth 10 times one-in-a-line. There is also a small bonus for disks in the centre column, because the centre takes part in the most lines. The values are kept well below a real win score (which is at least 1000) so a forced win always beats a merely good-looking position.

```python
WIN = 1000  # base value of a win; real wins score WIN + empty squares


def evaluate(board):
    """Heuristic score of a non-terminal position from 'o''s point of view."""
    score = 0
    cells = board.cells
    for line in board.lines:
        n_o = n_x = 0
        for r, c in line:
            v = cells[r][c]
            if v == "o":
                n_o += 1
            elif v == "x":
                n_x += 1
        if n_o and not n_x:
            score += 10 ** (n_o - 1)
        elif n_x and not n_o:
            score -= 10 ** (n_x - 1)
    centre = board.cols // 2
    for r in range(board.rows):
        if cells[r][centre] == "o":
            score += 2
        elif cells[r][centre] == "x":
            score -= 2
    return score
```

#### Minimax with alpha-beta, caching and symmetry

A note on caching with alpha-beta: when a branch is pruned, the value we got back is only a **bound**, not the exact value. For example, "this position is worth *at least* 5" is all we learn if we stopped early. So each cache entry stores a flag saying whether the value is `EXACT`, a `LOWER` bound or an `UPPER` bound, and we only reuse it when that is enough for the current alpha-beta window. If we only cached exact values, the code would still be correct but would cache much less.

Each entry also stores the remaining search depth. A value found by a shallow, depth-limited search must not be trusted when a deeper search is requested.

```python
EXACT, LOWER, UPPER = 0, 1, 2
OTHER = {"o": "x", "x": "o"}


class MinimaxAgent:
    def __init__(self, num_rows=3, num_cols=5, num_connect=3, max_depth=None):
        self.max_depth = max_depth          # None = search to the end of the game
        self.cache = {}                     # transposition table
        centre = (num_cols - 1) / 2
        self.order = sorted(range(num_cols), key=lambda c: abs(c - centre))  # centre-first
        self.nodes = 0

    def minimax(self, board, player, depth, alpha, beta):
        """Value of `board` with `player` to move (positive = good for 'o')."""
        self.nodes += 1
        alpha0, beta0 = alpha, beta

        # 1. Cache lookup (symmetry-aware)
        key, mirrored = board.key()
        entry = self.cache.get(key)
        best_hint = None
        if entry is not None:
            value, flag, d, best = entry
            if best is not None:
                best_hint = board.cols - 1 - best if mirrored else best
            if d is None or (depth is not None and d >= depth):
                if flag == EXACT:
                    return value
                if flag == LOWER:
                    alpha = max(alpha, value)
                elif flag == UPPER:
                    beta = min(beta, value)
                if alpha >= beta:
                    return value

        # 2. Draw (board full). Wins are detected when the move is made, below.
        if board.empty == 0:
            return 0

        # 3. Depth limit reached -> heuristic estimate
        if depth == 0:
            return evaluate(board)

        # 4. Move ordering: cached best move first, then centre-out
        moves = [c for c in self.order if board.heights[c] < board.rows]
        if best_hint in moves:
            moves.remove(best_hint)
            moves.insert(0, best_hint)

        maximising = (player == "o")
        best_value = -float("inf") if maximising else float("inf")
        best_move = None
        next_depth = None if depth is None else depth - 1

        for col in moves:
            r = board.play(col, player)
            if board.is_win_at(r, col, player):
                value = (WIN + board.empty) * (1 if maximising else -1)
            else:
                value = self.minimax(board, OTHER[player], next_depth, alpha, beta)
            board.undo(col)

            if maximising:
                if value > best_value:
                    best_value, best_move = value, col
                alpha = max(alpha, value)
            else:
                if value < best_value:
                    best_value, best_move = value, col
                beta = min(beta, value)
            if alpha >= beta:
                break  # prune: the opponent will never allow this position

        # 5. Store the result with the right flag
        if best_value <= alpha0:
            flag = UPPER
        elif best_value >= beta0:
            flag = LOWER
        else:
            flag = EXACT
        stored_best = board.cols - 1 - best_move if mirrored else best_move
        self.cache[key] = (best_value, flag, depth, stored_best)
        return best_value

    def move_values(self, env):
        """Exact (or depth-limited) value of every legal move in the current env position."""
        board = Board.from_env(env)
        player = env.active_player
        values = {}
        for col in board.legal_moves():
            r = board.play(col, player)
            if board.is_win_at(r, col, player):
                values[col] = (WIN + board.empty) * (1 if player == "o" else -1)
            else:
                depth = None if self.max_depth is None else self.max_depth - 1
                values[col] = self.minimax(board, OTHER[player], depth, -float("inf"), float("inf"))
            board.undo(col)
        return values

    def choose_move(self, env):
        values = self.move_values(env)
        pick = max if env.active_player == "o" else min
        best = pick(values.values())
        # Among equally good moves, prefer the most central one
        return next(c for c in self.order if values.get(c) == best)

    # --- Table lookup: save / load the cache ---
    def save_table(self, path):
        with open(path, "wb") as f:
            pickle.dump(self.cache, f)

    def load_table(self, path):
        with open(path, "rb") as f:
            self.cache = pickle.load(f)
```

### Solving Connect Three

With no depth limit the agent searches all the way to the end of the game, so its moves are **optimal**. Let's solve the empty board (with `'x'` moving first) and time it.

```python
agent = MinimaxAgent(max_depth=None)
env = connect.Connect(starting_player="x", verbose=False)

t = time.time()
values = agent.move_values(env)
print(f"Solved in {time.time() - t:.2f}s, {agent.nodes:,} nodes searched, {len(agent.cache):,} positions cached")
print("Value of each first move for 'x' (negative = good for x):", values)


def describe(v):
    if v == 0:
        return "draw"
    winner = "o" if v > 0 else "x"
    empty_left = abs(v) - WIN
    return f"'{winner}' wins with {empty_left} empty squares left"


for col, v in values.items():
    print(f"  x plays column {col}: {describe(v)}")
```

How much do the optimisations help? Here we switch them off one at a time and count the nodes searched to solve the empty board.

For comparison, **plain minimax with no optimisations searches about 19.5 million nodes and takes over a minute** for the same job. The optional cell after this one runs it if you want to see that for yourself.

```python
class NoSymmetryBoard(Board):
    def key(self):
        return "|".join("".join(row) for row in self.cells), False


def solve_stats(board_cls=Board, use_cache=True, centre_first=True):
    a = MinimaxAgent()
    if not centre_first:
        a.order = list(range(5))
    if not use_cache:
        a.cache = type("NoCache", (dict,), {"__setitem__": lambda self, k, v: None})()
    b = board_cls()
    t = time.time()
    v = a.minimax(b, "x", None, -float("inf"), float("inf"))
    return v, a.nodes, time.time() - t


for name, kwargs in [
    ("alpha-beta + cache + symmetry + centre-first", {}),
    ("no symmetry", {"board_cls": NoSymmetryBoard}),
    ("no centre-first move ordering", {"centre_first": False}),
    ("alpha-beta only (no cache)", {"use_cache": False}),
]:
    v, n, s = solve_stats(**kwargs)
    print(f"{name:48s} value={v:5}  nodes={n:>9,}  time={s:.2f}s")
```

```python
RUN_PLAIN = False  # set to True to run plain minimax (slow: about a minute)


def plain_minimax(board, player, counter):
    counter[0] += 1
    if board.empty == 0:
        return 0
    values = []
    for col in board.legal_moves():
        r = board.play(col, player)
        if board.is_win_at(r, col, player):
            v = (WIN + board.empty) * (1 if player == "o" else -1)
        else:
            v = plain_minimax(board, OTHER[player], counter)
        board.undo(col)
        values.append(v)
    return max(values) if player == "o" else min(values)


if RUN_PLAIN:
    counter = [0]
    t = time.time()
    v = plain_minimax(Board(), "x", counter)
    print(f"plain minimax: value={v}  nodes={counter[0]:,}  time={time.time() - t:.1f}s")
```

### Testing the AI

To test it, the perfect AI plays a random player 400 times, covering every combination of which side the AI takes and who moves first. We know from the solution that the first player can force a win. So when the AI moves first it should **win every game**. When the random player moves first, the AI can only lose if the random player happens to find the winning line, and it should still win most of those games. The AI also plays itself, and that game should end exactly as the solved value predicts: `'x'` wins with 6 empty squares left.

```python
def play_game(agent_o, agent_x, starting_player="x", verbose=False):
    """agent_* is a MinimaxAgent or the string 'random'. Returns the reward for 'o'."""
    env = connect.Connect(starting_player=starting_player, verbose=verbose)
    game_over = False
    while not game_over:
        a = agent_o if env.active_player == "o" else agent_x
        move = random.choice(list(env.available_actions)) if a == "random" else a.choose_move(env)
        reward, game_over = env.act(move)
    return reward


random.seed(0)
for ai_side in ["o", "x"]:
    for starter in ["o", "x"]:
        tally = {"AI wins": 0, "draws": 0, "AI loses": 0}
        for _ in range(100):
            if ai_side == "o":
                reward = play_game(agent, "random", starter)
            else:
                reward = play_game("random", agent, starter)
            ai_reward = reward if ai_side == "o" else -reward
            tally[{1: "AI wins", 0: "draws", -1: "AI loses"}[ai_reward]] += 1
        who = "AI" if starter == ai_side else "random"
        print(f"AI plays '{ai_side}', {who:6s} moves first: {tally}")

print("\nPerfect play vs perfect play (x starts):")
r = play_game(agent, agent, "x", verbose=True)
```

### Extension: full Connect Four (6 rows × 7 columns, connect 4)

The code is written for any board size, but Connect Four is far too big to solve this way in Python, so we turn on the **depth limit**. When the search reaches it, the **evaluation function** estimates the position. Below, a depth-6 agent plays a depth-2 agent, once with each side moving first. Each move takes a fraction of a second. The deeper searcher should never lose, although a draw is possible because both agents are imperfect.

```python
def make_c4_env(start):
    return connect.Connect(starting_player=start, num_cols=7, num_rows=6, num_connect=4, verbose=False)


def play_c4(agent_o, agent_x, start):
    env = make_c4_env(start)
    game_over, moves = False, 0
    while not game_over:
        a = agent_o if env.active_player == "o" else agent_x
        a.cache.clear()  # depth-limited values go stale as the game moves on; keep memory small
        reward, game_over = env.act(a.choose_move(env))
        moves += 1
    return reward, moves, env


for start in ["o", "x"]:
    deep = MinimaxAgent(6, 7, 4, max_depth=6)
    shallow = MinimaxAgent(6, 7, 4, max_depth=2)
    t = time.time()
    reward, moves, env4 = play_c4(deep, shallow, start)   # deep agent is 'o'
    outcome = {1: "depth-6 AI ('o') wins", -1: "depth-2 AI ('x') wins", 0: "draw"}[reward]
    print(f"'{start}' starts: {outcome} after {moves} moves ({time.time() - t:.1f}s)")
    print(env4.grid[::-1], "\n")
```

### Reflection

* **Connect Three (3×5) is solved exactly.** The table above shows that with perfect play the first player wins by starting in any of the three middle columns, and loses if they start on an edge. When the AI moves first it wins every game against a random player. When it moves second it occasionally loses, but only because the random player happened to find the forced win.
* **Alpha-beta and the cache do most of the work.** Alpha-beta skips branches the opponent would never allow. The cache stops the search re-solving positions reached by different move orders, which is very common in Connect games because the order of moves in different columns often doesn't matter. Symmetry roughly halves the cache, and centre-first move ordering finds good moves early, so more branches get pruned.
* **Caching with alpha-beta needs care.** A pruned search only gives a bound, so entries carry an `EXACT`/`LOWER`/`UPPER` flag. Without that the AI would sometimes reuse a wrong value and play badly.
* **Connect Four needs a depth limit**, which means giving up guaranteed optimal play. How strong the AI is then depends on the evaluation function. Possible improvements are iterative deepening (search depth 1, 2, 3, … until a time limit runs out), better move ordering from earlier iterations, and a bitboard representation for speed.
