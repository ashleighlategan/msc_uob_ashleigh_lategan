### Objective
You should write or adapt the minimax algorithm from the noughts and crosses example to work for connect three. The state space is much larger due to the size of the board, so the basic brute force version of the algorithm is unlikely to be computationally efficient enough to actually play against. Hence you will be required to implement some optimisations, for example:
* Alpha-beta pruning – see the noughts and crosses example.
* Cache results for board states – multiple paths can lead to the same state so you shouldn't have to solve these twice. You can use a dictionary to cache results efficiently.
* Optimise for symmetry – the board is symmetric around the centre column, so you can cut down the number of necessary states by assuming that optimal play will look the same but adjusted for reflection when you are creating the cache.
* Table lookups – like caching results but saving permanently, you can precompute a certain number of end-game board positions and store the results (e.g. calculate in one cell and then use in another).
* Depth limit – a hard cap on the recursion limit will hugely improve performance at the cost of optimality. If the recursion limit is reached you can simply return a default value like 0, or...
* Evaluation functions – rather than returning a default value when you hit the recursion limit, you can use a fast heuristic function to evaluate the strength of a board position for the given player. What you choose is up to you! Be creative!

```python
import connect
import numpy as np
import random
import time
import pickle

class Board:
    """
    Fast, mutable board used inside the search.

    Row 0 is at the bottom, as in Connect.
    """

    def __init__(self, num_rows=3, num_cols=5, num_connect=3):
        self.rows, self.cols, self.k = num_rows, num_cols, num_connect
        self.cells = [[" "] * num_cols for _ in range(num_rows)]
        self.heights = [0] * num_cols          # number of disks in each column
        self.empty = num_rows * num_cols       # empty cells left
        self.lines = self._all_lines()         # every possible line of k cells i.e. all possible ways to win

        # Lookup table for each cell to see which winning lines include that cell. 
        # e.g. lines_through[(0, 0)] will give the 3 winning lines that use the bottom-left corner.
        # After a move we only need to check the lines through the cell just played,
        # because a new disk can't complete a line it isn't part of (avoids wasted effort). 

        # create a dictionary with one entry per cell,
        # each holding an empty list to be filled with that cell's winning lines
        self.lines_through = {(r, c): [] for r in range(num_rows) for c in range(num_cols)}
        # Now add each winning line to the list of every cell it includes. 

        for line in self.lines:                         # line = one winning line, e.g. [(0,0), (0,1), (0,2)]
            for sq in line:                             # sq = one cell in that line, e.g. (0,0)    
                self.lines_through[sq].append(line)     # add this line to that cell's list

    # Classmethod: builds a new Board that copies the current state of the game 
    # in `env`, and returns it all in a single step (advantage of using Classmethod, you dont have to create a board first).
    @classmethod
    def from_env(cls, env):
        """
        Return a new Board that copies the current game in env.

        Used by the AI to get its own copy to search on, so trying out
        moves never changes the real game. Usage: Board.from_env(env)
        """
        b = cls(env._num_rows, env._num_cols, env._num_connect)
        for r in range(b.rows):
            for c in range(b.cols):
                if env.grid[r, c] != " ":
                    b.cells[r][c] = str(env.grid[r, c])
                    b.heights[c] += 1       # counts the disks in each column
                    # counts the empty cells left on the board
                    # Starts at 15 (3 × 5), and each disk found takes 1 away.
                    b.empty -= 1            
        return b

    def _all_lines(self):
        """
        Return all possible winning lines on the board.

        Each line is a list of k (row, col) cells, e.g. [(0, 0), (0, 1), (0, 2)].
        Checks 4 directions from every cell (right, up, up-right diagonal, up-left diagonal)
        and keeps only the lines that fit on the board.
        """
        lines = []
        for r in range(self.rows):
            for c in range(self.cols):
                for dr, dc in [(0, 1), (1, 0), (1, 1), (1, -1)]:  # right, up, up-right, up-left
                    line = [(r + i * dr, c + i * dc) for i in range(self.k)]
                    if all(0 <= rr < self.rows and 0 <= cc < self.cols for rr, cc in line):
                        lines.append(line)
        return lines

    def legal_moves(self):
        """Return the columns that aren't populated yet."""
        return [c for c in range(self.cols) if self.heights[c] < self.rows]

    def play(self, col, player):
        """Drop player's disk into column (col) and return the row it lands in."""
        r = self.heights[col]
        self.cells[r][col] = player     # player here is the disk, either x or o so write this player's letter into the cell at row r, column col
        # adds 1 to the number of disks in that one column
        self.heights[col] += 1
        # takes 1 away from the count of empty cells on the whole board
        self.empty -= 1
        return r

    def undo_move(self, col):
        """Remove the top disk from column col (undoes the last play)."""
        self.heights[col] -= 1
        self.cells[self.heights[col]][col] = " "
        self.empty += 1

    def is_win_at(self, r, c, player):
        """Did the disk just placed at (r, c) complete a line for player, such that the game is over?"""
        cells = self.cells
        # for each cell (rr, cc) in this line, does it contain this player's letter?
        # Then run the inner check on each line through the cell that was just played, and asks if any of them came back True (a win).
        return any(all(cells[rr][cc] == player for rr, cc in line)
                   for line in self.lines_through[(r, c)])

    def key(self):
        """
        Return the cache key for this board.

        A board and its mirror image (flipped left to right) are equally
        good positions, so we only want to solve one of them and save it once to avid wasted effort.

        We do this by writing the board out as text, and also its mirror image, then
        pick whichever comes first alphabetically. Both boards pick the same
        one, so they share one cache entry. For example, these two boards
        get the same key:

            x o . . .        . . . o x
            (board)          (mirror)

        Returns:
            key:       The text to look up in the cache
            mirrored:  True if the key came from the mirror image. If so, any
                       best move saved in the cache is also mirrored, so it
                       must be flipped back (column 0 <-> 4, 2 <-> 2, 1 <-> 3, ).
        """
        k = "|".join("".join(row) for row in self.cells)
        k_mirror = "|".join("".join(reversed(row)) for row in self.cells)
        return (k, False) if k <= k_mirror else (k_mirror, True)

    def __str__(self):
        return str(np.array(self.cells)[::-1])


```

```python
# evaluation function
# When the depth limit is reached we can't see the end of the game, so we need to estimate how good the position is for 'o'.
# For every possible line of k cells:
            # if it contains only 'o' disks (and blanks), it is still playable for `'o'`, so add `10^(n-1)` where n is the number of 'o' disks in it;
            # if it contains only 'x' disks, subtract the same amount
            # if it contains both 'o' and 'x' it is dead and counts for nothing.

WIN = 1000  # base value of a win; real wins score WIN + empty cells

def evaluate(board):
    """Heuristic score of a non-terminal position from the perspective of player 'o'"""
    score = 0
    cells = board.cells
    for line in board.lines:
        n_o = n_x = 0
        for r, c in line:
            v = cells[r][c]
            if v == "o":
                n_o += 1
            elif v == "x":
                n_x += 1
        if n_o and not n_x:
            score += 10 ** (n_o - 1)
        elif n_x and not n_o:
            score -= 10 ** (n_x - 1)
    centre = board.cols // 2
    for r in range(board.rows):
        if cells[r][centre] == "o":
            score += 2
        elif cells[r][centre] == "x":
            score -= 2
    return score
```

```python
# Minimax with alpha-beta pruning, caching and symmetry
# Since with pruning, search can stop early so each cache entry gets a label:
            # EXACT - the true score   
            # LOWER - the true score is at least this e.g. it stopped searching at 50: so at least 50
            # UPPER - the true score is at most this e.g. it stopped once it couldn't reach a pass: at most 40

EXACT, LOWER, UPPER = 0, 1, 2
OTHER = {"o": "x", "x": "o"}


class MinimaxAgent:
    def __init__(self, num_rows=3, num_cols=5, num_connect=3, max_depth=None):
        self.max_depth = max_depth          # None = search to the end of the game
        self.cache = {}                     # transposition table: dictionary of board key -> saved result
        centre = (num_cols - 1) / 2
        self.order = sorted(range(num_cols), key=lambda c: abs(c - centre))  # centre-first
        self.nodes = 0

    def minimax(self, board, player, depth, alpha, beta):
        """
        Work out how good this board is, given whose turn it is.

        The result is a score where positive means good for 'o'.
        """
        self.nodes += 1
        alpha0, beta0 = alpha, beta         # note down the starting values

        # 1. Cache lookup (symmetry-aware)
        key, mirrored = board.key()
        entry = self.cache.get(key)
        best_hint = None
        if entry is not None:
            value, flag, d, best = entry
            if best is not None:
                best_hint = board.cols - 1 - best if mirrored else best
            if d is None or (depth is not None and d >= depth):
                if flag == EXACT:
                    return value
                if flag == LOWER:
                    alpha = max(alpha, value)
                elif flag == UPPER:
                    beta = min(beta, value)
                if alpha >= beta:
                    return value

        # 2. Draw (board full). Wins are detected when the move is made, below.
        if board.empty == 0:
            return 0

        # 3. Depth limit reached -> heuristic estimate
        # If we've looked as far ahead as we're allowed, stop searching and make an educated guess about the board instead.
        if depth == 0:
            return evaluate(board)

        # 4. Order moves best-guess first:
        #    - the best move saved in the cache for this board (if any)
        #    - then the remaining columns, centre first, edges last
        #    Good moves early let alpha-beta pruning skip more of the bad ones.

        # List the legal moves, centre first
        # The centre is usually the strongest column because it's part of the most winning lines, so it's tried first
        moves = [c for c in self.order if board.heights[c] < board.rows]

        # put the cache's best move at the front
        if best_hint in moves:
            moves.remove(best_hint)
            moves.insert(0, best_hint)

        maximising = (player == "o")
        best_value = -float("inf") if maximising else float("inf")
        best_move = None
        next_depth = None if depth is None else depth - 1

        for col in moves:
            r = board.play(col, player)
            if board.is_win_at(r, col, player):
                value = (WIN + board.empty) * (1 if maximising else -1)
            else:
                value = self.minimax(board, OTHER[player], next_depth, alpha, beta)
            board.undo_move(col)

            if maximising:
                if value > best_value:
                    best_value, best_move = value, col
                alpha = max(alpha, value)
            else:
                if value < best_value:
                    best_value, best_move = value, col
                beta = min(beta, value)
            if alpha >= beta:
                break  # prune: the opponent will never allow this position

        # 5. Store the result with the right flag
        # Alpha-beta may have stopped searching early, so the result might only be a limit, not the true score.
        # Note the search only ever stops early when it sees a score outside the window. That's the only reason it ever stops.

        # This is no better than what 'o' already had elsewhere, so the search was cut short.
        # True score is AT MOST this.
        # A branch is only worth exploring if it could beat alpha. 
        # As soon as 'x' shows that a branch will end at or below alpha, 'o' stops looking at it.
        if best_value <= alpha0:
            flag = UPPER
        # 'x' would never allow this, so the search was cut short.
        # True score is AT LEAST this.
        elif best_value >= beta0:
            flag = LOWER
        # Landed somewhere between alpha and beta: nothing was cut short.
        # This is the TRUE score.
        else:
            flag = EXACT
        stored_best = board.cols - 1 - best_move if mirrored else best_move
        self.cache[key] = (best_value, flag, depth, stored_best)
        return best_value

    def move_values(self, env):
        """
        Exact (or depth-limited) value of every legal move in the current env position.
        
        For each move I could make right now, what score would it lead to?
        """
        board = Board.from_env(env)         # copy the real game onto the AI's own board
        player = env.active_player          # whose turn is it? 'o' or 'x' ?
        values = {}                         # empty dictionary to fill: column -> score

        for col in board.legal_moves():     # try each column that isn't full
            r = board.play(col, player)     # drop the disk (r = row it landed in)

            # Did that move win straight away?
            if board.is_win_at(r, col, player):
                values[col] = (WIN + board.empty) * (1 if player == "o" else -1)
            else:
                # if not a straight away win, check with minimax what happens next
                depth = None if self.max_depth is None else self.max_depth - 1
                values[col] = self.minimax(board, OTHER[player], depth, -float("inf"), float("inf"))
            board.undo_move(col)           # take the disk back out before trying the next column
        return values                      # return all columns and their scores

    def choose_move(self, env):
        """
        Return the best column to play in the current game based on the scores from move_values.

        'o' picks the highest-scoring move, 'x' picks the lowest.
        If several moves tie, the most central one is chosen as playing in the center gives you 
        more ways to win if the opponent slips up.
        """
        values = self.move_values(env)
        pick = max if env.active_player == "o" else min
        best = pick(values.values())                    # values.values() gives just the scores, without the columns
        # Among equally good moves, prefer the most central one
        # self.order is already sorted centre first
        # so the line just takes the first good move it finds, and the list is arranged so that most central are checked first.
        return next(c for c in self.order if values.get(c) == best)

    # --- Table lookup: save / load the cache ---
    def save_table(self, path):
        """
        Save the cache to a file, so solved positions can be reused later.

        Solve the game once and save the cache to a file.
        Next time, load the file, and every position is already solved, so each move is just a lookup
        How to use: agent.save_table("connect3_table.pkl")
        """
        with open(path, "wb") as f:         # open (or create) the file for Writing, in Binary
            pickle.dump(self.cache, f)      # save the whole dictionary into it

    def load_table(self, path):
        """
        Load a saved cache from a file, replacing the current cache.

        Positions in the file are then looked up instead of searched again.
        How to use: agent.load_table("connect3_table.pkl")
        """
        with open(path, "rb") as f:         # open the file for Reading, in Binary
            self.cache = pickle.load(f)     # read the dictionary back and use it as the cache
```

```python
# Baseline: plain minimax with NO optimisations 
# No alpha-beta, no cache, no symmetry, no move ordering.
# Searches every possible game to the end, so it is slow ~ a minute.
# Compare its node count with the optimised versions below.

RUN_PLAIN = True  # set to True to run plain minimax (slow: about a minute)


def plain_minimax(board, player, counter):
    """
    Return the score of board with player to move, using basic minimax.

    Tries every legal move and searches every reply to the end of the game.
    Score are positive if 'o' wins, negative if 'x' wins and 0 if there is a draw.

    counter: a one-item list, [number of positions searched], shared by all calls
             (a list is used so every recursive call adds to the same count)
             This works as a list , unlike a value, isn't copied when you pass it to a function. 
             The function gets the same list, so changing what's inside it changes it for all.
    """
    counter[0] += 1                                     # count this position
    if board.empty == 0:                                # board full and nobody won so it's a draw
        return 0
    values = []                                         # score of each possible move
    for col in board.legal_moves():                     # try every move (no pruning)
        r = board.play(col, player)
        if board.is_win_at(r, col, player):             # this move wins straight away
            v = (WIN + board.empty) * (1 if player == "o" else -1)
        else:                                           # no win right away, so see how the opponent replies
            v = plain_minimax(board, OTHER[player], counter)
        board.undo_move(col)                            # take the move back
        values.append(v)

    # 'o' picks the highest score, 'x' picks the lowest
    return max(values) if player == "o" else min(values)


if RUN_PLAIN:
    counter = [0]                                       # positions searched so far
    t = time.time()                                     # start stopwatch
    v = plain_minimax(Board(), "x", counter)            # solve from the empty board, 'x' to move
    print(f"plain minimax: value={v}  nodes={counter[0]:,}  time={time.time() - t:.1f}s")
```

```python
# Here we solve Connect Three several times, each time with one optimisation switched off
# and compare the effort needed to solve the game


# A board with symmetry switched off
# subclass: it inherits everything from Board but replaces key()
# new key() just returns the board as text and never checks the mirror.
# so a board and its mirror get different keys and are solved separately.
class NoSymmetryBoard(Board):
    """A Board that ignores symmetry: a board and its mirror get different cache keys."""
    def key(self):
        return "|".join("".join(row) for row in self.cells), False

class NoCache(dict):
    """A 'cache' that never stores anything, so no position is ever reused."""

    def __setitem__(self, key, value):
        pass  # ignore every attempt to save, so the cache stays empty

# How much work does it take to solve Connect Three with these optimisations switched on or off?
def solve_stats(board_cls=Board, use_cache=True, centre_first=True):
    """
    Solve Connect Three from the empty board with some optimisations switched off.

    Returns (value, nodes searched, seconds taken).
    """
    a = MinimaxAgent()
    if not centre_first:
        a.order = list(range(5))     # try columns 0, 1, 2, 3, 4 instead of centre first
    if not use_cache:
        a.cache = NoCache()          # swap in a cache that can't remember anything
    b = board_cls()
    t = time.time()
    v = a.minimax(b, "x", None, -float("inf"), float("inf"))
    return v, a.nodes, time.time() - t

# runs the 4 experiments one after another and prints a results row for each\
# kwargs holds the settings for each experiment 
for name, kwargs in [
    ("alpha-beta + cache + symmetry + centre-first", {}),                # 1. everything on
    ("no symmetry", {"board_cls": NoSymmetryBoard}),                     # 2. symmetry off
    ("no centre-first move ordering", {"centre_first": False}),          # 3. centre-first off
    ("alpha-beta only (no cache)", {"use_cache": False}),                # 4. cache off
]:
    
    # '**' unpacks a dictionary into named arguments. 
    # It takes each "name": value pair from the dictionary and passes it to the function as name=value.
    v, n, s = solve_stats(**kwargs)                     
    print(f"{name:48s} value={v:5}  nodes={n:>9,}  time={s:.2f}s")
```

```python
# Testing the AI
# Check how the AI plays by playing lots of real games against a random player.
# What perfect play should give:
#   - AI moves first      -> AI wins EVERY game (the first player can force a win)
#   - random moves first  -> AI wins most games; it can only lose if the random player
#                            happens to stumble onto the forced win
# Any loss when the AI moves first would mean there's a bug.

def play_game(agent_o, agent_x, starting_player="x", verbose=False):
    """
    Play one full game and return the result.

    agent_o, agent_x: the player for each side, either a MinimaxAgent or the string "random"
    starting_player:  who moves first, 'o' or 'x'
    verbose:          if True, print the board after every move

    Returns the reward for 'o': +1 = 'o' wins, -1 = 'x' wins, 0 = draw.
    """
    env = connect.Connect(starting_player=starting_player, verbose=verbose)   # new game
    game_over = False
    while not game_over:
        a = agent_o if env.active_player == "o" else agent_x                  # whose turn?
        if a == "random":
            move = random.choice(list(env.available_actions))                 # pick any legal column
        else:
            move = a.choose_move(env)                                         # ask AI's for its best move
        reward, game_over = env.act(move)                                     # play it on the real board
    return reward


# Play a total of 400 games: 4 set-ups (AI as 'o' or 'x', and either side moving first),
# with 100 games of each
random.seed(0)  # fix the random player's choices so results are the same every run

for ai_side in ["o", "x"]:                 # AI plays as 'o', then as 'x'
    for starter in ["o", "x"]:             # 'o' moves first, then 'x' moves first
        tally = {"AI wins": 0, "draws": 0, "AI loses": 0}     # scoreboard for this combination

        for _ in range(100):               # play 100 games
            if ai_side == "o":
                reward = play_game(agent, "random", starter)
            else:
                reward = play_game("random", agent, starter)

            # reward is from 'o''s point of view; flip it if the AI is 'x' so +1 = AI won
            ai_reward = reward if ai_side == "o" else -reward
            result = {1: "AI wins", 0: "draws", -1: "AI loses"}[ai_reward]
            tally[result] += 1

        who = "AI" if starter == ai_side else "random"
        print(f"AI plays '{ai_side}', {who:6s} moves first: {tally}")


# AI vs AI:
# Both sides play perfectly, so this should match the solve: 'x' wins with 6 empty cells left.
print("\nPerfect play vs perfect play (x starts):")
r = play_game(agent, agent, "x", verbose=True)              # AI is both 'o' AND 'x'
```

```python
# Table lookup: save the solved game so we can reuse it 
# agent has already solved the whole game, so its cache holds every position it worked out.
# Save that cache to a file, load it into a brand-new AI, and show it can now pick moves
# almost instantly, by looking positions up instead of searching.

agent.save_table("connect3_table.pkl")          # save the solved positions to a file

fast_agent = MinimaxAgent()                     # a new AI with an empty cache...
fast_agent.load_table("connect3_table.pkl")     # ...that loads all the saved positions

env.reset()                                     # clear the board to time the opening move
t = time.time()                                 # start the stopwatch
move = fast_agent.choose_move(env)

# Should be fast ~ under a millisecond, with only 5 nodes searched
# (one lookup for each of the 5 possible first moves)
print(f"Opening move {move} chosen in {1000 * (time.time() - t):.1f} ms using the saved table "
      f"({fast_agent.nodes} new nodes searched)")
```

```python
# What if you want to try play against the AI?
# Set PLAY_NOW = True and run this cell. Type a column number when asked.
# Columns are numbered 0-4 from left to right.

def play_vs_human(agent, human="x", starting_player="x"):
    """
    Play a game of Connect Three against the AI, typing your moves in.

    agent:           the AI opponent (a MinimaxAgent)
    human:           which side you play, 'o' or 'x' (the AI plays the other)
    starting_player: who moves first, 'o' or 'x'

    The board is printed after every move.
    """
    env = connect.Connect(starting_player=starting_player, verbose=True)   # new game, prints the board
    game_over = False

    while not game_over:
        if env.active_player == human:
            # Your turn - it will keep asking you until you type a valid column
            move = None
            while move not in env.available_actions:
                try:
                    move = int(input(f"Your move ({human}), choose from {list(env.available_actions)}: "))
                except ValueError:
                    pass   # you typed something that isn't a number, so ask again
        else:
            # AI's turn
            move = agent.choose_move(env)
            print(f"AI ({env.active_player}) plays column {move}")

        reward, game_over = env.act(move)   # play the move on the board

    # reward is from 'o''s point of view: +1 is if 'o' won, -1 is if 'x' won and 
    # 0 is in the case of a draw
    if reward == 0:
        print("Draw!")
    elif (reward == 1 and human == "o") or (reward == -1 and human == "x"):
        print("You win!")
    else:
        print("The AI wins!")

# set to True to play (it waits for your input, so it's off by default)
PLAY_NOW = True  
if PLAY_NOW:
    play_vs_human(fast_agent, human="x", starting_player="x")   # you are 'x' and move first
```

```python
# What changes are needed to extend this to connect 4 (6 rows × 7 columns)?
# Connect Four is too big to search to the end in Python, so both AIs use a DEPTH LIMIT:
# They look a fixed number of moves ahead, then the evaluation function estimates the position.
# Here a depth-6 AI (looks 6 moves ahead) plays a depth-2 AI (looks 2 moves ahead),
# once with each side moving first. The deeper AI should never lose.

def make_c4_env(start):
    """
    Return a new Connect Four game (6 rows, 7 columns, 4 in a row to win).

    start: who moves first, 'o' or 'x'
    """
    return connect.Connect(starting_player=start, num_cols=7, num_rows=6, num_connect=4, verbose=False)


def play_c4(agent_o, agent_x, start):
    """
    Play one game of Connect Four between two AIs.

    agent_o, agent_x: the MinimaxAgent playing each side
    start:            who moves first, 'o' or 'x'

    Returns (reward, number of moves played, the finished game).
    reward is from 'o''s point of view: +1 = 'o' wins, -1 = 'x' wins, 0 = draw.
    """
    env = make_c4_env(start)
    game_over, moves = False, 0
    while not game_over:
        a = agent_o if env.active_player == "o" else agent_x   # whose turn?
        # Empty the cache before each move to keep memory small. Positions from earlier
        # in the game won't come up again (disks are never removed from the real board).
        a.cache.clear()
        reward, game_over = env.act(a.choose_move(env))        # AI picks a move and plays it
        moves += 1
    return reward, moves, env


# --- Depth-6 AI vs depth-2 AI, once with each side moving first ---
for start in ["o", "x"]:
    deep = MinimaxAgent(6, 7, 4, max_depth=6)      # rows, cols, in-a-row, looks 6 moves ahead
    shallow = MinimaxAgent(6, 7, 4, max_depth=2)   # looks only 2 moves ahead
    t = time.time()                                # start the stopwatch
    reward, moves, env4 = play_c4(deep, shallow, start)   # deep AI is 'o', shallow AI is 'x'

    outcome = {1: "depth-6 AI ('o') wins", -1: "depth-2 AI ('x') wins", 0: "draw"}[reward]
    print(f"'{start}' starts: {outcome} after {moves} moves ({time.time() - t:.1f}s)")
    print(env4.grid[::-1], "\n")                   # final board, flipped so the bottom row prints last
```

### Takeaways

* **Connect Three (3×5) is solved exactly.** The table above shows that with perfect play the first player wins by starting in any of the three middle columns, and loses if they start on an edge. When the AI moves first it wins every game against a random player. When it moves second it occasionally loses, but only because the random player happened to find the forced win.

* **Alpha-beta and the cache do most of the work.** Alpha-beta skips branches the opponent would never allow. The cache stops the search re-solving positions reached by different move orders, which is very common in Connect games because the order of moves in different columns often doesn't matter. Symmetry roughly halves the cache, and centre-first move ordering finds good moves early, so more branches get pruned.

* **Caching with alpha-beta needs care.** A pruned search only gives a bound, so entries carry an `EXACT`/`LOWER`/`UPPER` flag. Without the labels, the AI would treat a limit like "at most 30" as an exact score of 30,reuse it later in a situation where the difference matters, and pick a worse move.

* **Connect Four needs a depth limit**, which means giving up guaranteed optimal play. How strong the AI is then depends on the 'evaluation function'. Possible improvements are iterative deepening (search depth 1, 2, 3, … until a time limit runs out), better move ordering from earlier iterations, and a bitboard representation for speed. Computers can process all the bits of a number at once, in one step. Checking for 4 in a row becomes a few arithmetic operations on whole numbers, instead of loops over rows and columns.
